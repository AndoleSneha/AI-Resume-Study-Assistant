```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# Load embedding model
embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Split text manually (no LangChain)
def split_text(text, chunk_size=300, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start = end - overlap
    return chunks

docs = split_text(my_profile)

# Create embeddings
embeddings = embedder.encode(docs)

# Build FAISS index
dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings).astype("float32"))

print("Memory chunks:", len(docs))
```

```python
from faster_whisper import WhisperModel
import numpy as np
import soundfile as sf
import tempfile
import io

whisper = WhisperModel("base")

def speech_to_text(audio_input):
    """
    Handles Gradio audio formats:
    - bytes
    - tuple (sr, numpy array)
    - numpy array
    - file path (str)
    """

    if audio_input is None:
        return ""

    tmp = tempfile.NamedTemporaryFile(suffix=".wav", delete=False)

    try:
        # Case 1: bytes
        if isinstance(audio_input, (bytes, bytearray)):
            audio_bytes = io.BytesIO(audio_input)
            data, sr = sf.read(audio_bytes)
            sf.write(tmp.name, data, sr)

        # Case 2: (sr, np.array)
        elif isinstance(audio_input, tuple):
            sr, data = audio_input
            sf.write(tmp.name, data, sr)

        # Case 3: numpy array
        elif isinstance(audio_input, np.ndarray):
            sf.write(tmp.name, audio_input, 16000)

        # Case 4: filepath
        elif isinstance(audio_input, str):
            return transcribe_path(audio_input)

        else:
            return ""

        segments, _ = whisper.transcribe(tmp.name)
        return " ".join([seg.text for seg in segments if seg.text])

    except Exception as e:
        return ""

def transcribe_path(path):
    segments, _ = whisper.transcribe(path)
    return " ".join([seg.text for seg in segments if seg.text])
```

```python
def get_memory(query):
    query_vec = embedder.encode([query]).astype("float32")
    D, I = index.search(query_vec, k=1)
    return docs[I[0][0]]
```

```python
from tavily import TavilyClient
from langchain_groq import ChatGroq
import os
import re

# Tavily
client = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])

# Groq LLM
llm = ChatGroq(
    api_key=os.environ["GROQ_API_KEY"],
    model="openai/gpt-oss-safeguard-20b",
    temperature=0.3
)

print("Brain & Tools Ready")

# --- TOOLS ---
def calculator_tool(text):
    try:
        return str(eval(text))
    except:
        return None

def web_search_tool(query):
    results = client.search(query, max_results=3)
    return str(results)

# --- ROUTER ---
def route_query(user_input):
    # If math detected
    if re.search(r"[0-9]+[\+\-\*/\(\)]", user_input):
        result = calculator_tool(user_input)
        if result:
            return f"🧮 Calculation Result: {result}"

    # If search intent detected
    search_keywords = ["search", "latest", "find", "news", "internship", "jobs"]
    if any(word in user_input.lower() for word in search_keywords):
        return f"🌐 Web Search Results:\n{web_search_tool(user_input)}"

    return None
```

```python
def digital_twin_chat(user_input):
    try:
        # If Gradio passed a tuple or mic array
        if not isinstance(user_input, str):
            user_input = speech_to_text(user_input)

        # Now user_input is text — chat
        tool_response = route_query(user_input)
        if tool_response:
            return tool_response

        memory = get_memory(user_input)

        prompt = f"""
You are my Digital Twin AI.

Memory:
{memory}

User:
{user_input}

Reply as a smart, confident, helpful version of me.
"""

        response = llm.invoke(prompt)
        return response if isinstance(response, str) else response.content

    except Exception as e:
        return f"⚠️ Internal Error: {str(e)}"
```

```python
from pypdf import PdfReader

def read_file(file_input):
    # Gradio gives a file path (string)
    path = file_input if isinstance(file_input, str) else file_input.name

    text = ""
    if path.lower().endswith(".pdf"):
        reader = PdfReader(path)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    else:
        with open(path, "r", encoding="utf-8", errors="ignore") as f:
            text = f.read()

    return text
```

```python
import gradio as gr

with gr.Blocks(css=custom_css, theme=gr.themes.Soft()) as ui:
    gr.Markdown("##My Digital Twin AI\nVoice, Chat & File-Based Smart Assistant")

    with gr.Row():
        file_input = gr.File(
            label="📁 Upload Resume / PDF",
            file_types=[".pdf", ".txt"]
        )

        text_input = gr.Textbox(
            label="💬 Type Your Message",
            placeholder="Ask me anything about coding, career, or life...",
            lines=2
        )

        audio_input = gr.Audio(
            sources=["microphone"],
            type="filepath",
            label="🎤 Speak",
        )

    output_box = gr.Textbox(
        label="🤖 Digital Twin Response",
        lines=8
    )

    def unified_input_handler(file, text, audio):

        # 1️⃣ Voice first
        try:
            if audio is not None:
                voice_text = speech_to_text(audio)
                if voice_text.strip():
                    return digital_twin_chat(voice_text)
        except:
            pass

        # 2️⃣ Text second
        if isinstance(text, str) and text.strip():
            return digital_twin_chat(text)

        # 3️⃣ File last
        if file is not None:
            try:
                file_text = read_file(file)
                if file_text.strip():
                    docs.append(file_text)
                    return "📄 File added to memory! Now ask me about it using voice or text."
            except Exception as e:
                return f"⚠️ File Error: {str(e)}"

        return "⚠️ Please speak, type, or upload a file before clicking Ask."

    submit_btn = gr.Button("🚀 Ask Digital Twin")

    submit_btn.click(
        fn=unified_input_handler,
        inputs=[file_input, text_input, audio_input],
        outputs=output_box
    )

ui.launch(share=True)
```
